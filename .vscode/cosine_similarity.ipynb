```python
import numpy as np
docs = np.array([
    [0.3, -1.2, 0.5, 0.9],
    [0.7, -2.7, 0.6, 0.1],
    [0.3, 0.01, 0.47, 2],
    [0.5, 2.2, 0.11, 3.0],
    [0.12, -0.7, 0.15, 0.25]
])
query = np.array([0.6, -1.5, 0.3, 0.2])
query_length = np.linalg.norm(query)
query_length
```

```python
q_norm = query / query_length
q_norm
```

```python
q_norm_length_test = np.linalg.norm(q_norm)
q_norm_length_test
```

```python
docs_lengths = np.linalg.norm(docs, axis=1, keepdims=True)
docs_lengths
```

```python
docs_norm = docs / docs_lengths
docs_norm
```

```python
docs_norm_test = np.linalg.norm(docs_norm, axis=1, keepdims=True)
docs_norm_test
```

```python
similarities = docs_norm @ q_norm
similarities
```

```python
best_ind_doc = np.argmax(similarities)
best_ind_doc
```
