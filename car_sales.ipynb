```python
import pandas as pd
df = pd.read_csv('car_sales_cleaned1.csv')
df
```

```python
df = df.drop_duplicates() #removing duplicates
df.shape
```

```python
df = df.dropna() #removing rows with missing values
df.shape
```

```python
valuesCol = df['Model'].unique()
```

```python
dfProjected = df[['Company', 'Model']]
dfProjected
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import converter
model = DecisionTreeClassifier()
y = dfProjected[['Company']]
mapper = converter.columnsMapper(['Model', 'Company'], dfProjected)
dfConverted = converter.convertX(dfProjected, mapper)
X = dfConverted[['Company']]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.8)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
score = accuracy_score(y_test, predictions)
score


```
