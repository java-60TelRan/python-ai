```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import converter
import joblib
df = pd.read_csv('car_sales_cleaned1.csv')

df = df.drop_duplicates() #removing duplicates
df = df.dropna() # removing rows with missing data
```

```python
model = DecisionTreeClassifier()
y = df[['Gender']]
mapper = converter.columnsMapper(['Gender','Company', 'Model', 'Transmission',	'Color'], df)
dfConverted = converter.convertX(df, mapper)
X = dfConverted[['Company', 'Model', 'Transmission' ]]
model.fit(X, y)
joblib.dump(model, 'cars-gender-model.joblib')

```
