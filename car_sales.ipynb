```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import converter
df = pd.read_csv('car_sales_cleaned1.csv')

df = df.drop_duplicates() #removing duplicates
df = df.dropna() # removing rows with missing data
```

```python
model = DecisionTreeClassifier()
y = df[['Model']]
mapper = converter.columnsMapper(['Gender','Company', 'Model', 'Transmission',	'Color'], df)
dfConverted = converter.convertX(df, mapper)
X = dfConverted[['Gender', 'Color' ]]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
score = accuracy_score(y_test, predictions)
score

```
